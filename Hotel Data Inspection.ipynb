```python

df["total_nights"] = (
    df["stays_in_weekend_nights"] +
    df["stays_in_week_nights"]
)

print("Total nights summary:")
print(df["total_nights"].describe())

print("\nAverage total nights:", round(df["total_nights"].mean(), 2))
print("Median total nights:", df["total_nights"].median())
print("Zero-night reservations:", (df["total_nights"] == 0).sum())
```

```python

long_stays = df[df["total_nights"] > 30]

print("Bookings longer than 30 nights:", len(long_stays))

print("\nHotel type:")
print(long_stays["hotel"].value_counts())

print("\nReservation status:")
print(long_stays["reservation_status"].value_counts())

print("\nStay details:")
print(
    long_stays[
        [
            "hotel",
            "total_nights",
            "adults",
            "children",
            "adr",
            "reservation_status",
            "market_segment"
        ]
    ].sort_values("total_nights", ascending=False).head(10)
)
```

```python

print("Most common stay lengths:")
print(df["total_nights"].value_counts().sort_index().head(15))

print("\nStay-length categories:")

stay_categories = pd.cut(
    df["total_nights"],
    bins=[-1, 0, 3, 7, 14, 30, float("inf")],
    labels=[
        "0 nights",
        "1–3 nights",
        "4–7 nights",
        "8–14 nights",
        "15–30 nights",
        "More than 30 nights"
    ]
)

print(stay_categories.value_counts().sort_index())
```

```python

stay_by_hotel = df.groupby("hotel")["total_nights"].agg(
    ["count", "mean", "median"]
).round(2)

print("Stay length by hotel type:")
print(stay_by_hotel)

print("\nStay categories by hotel type:")
print(
    pd.crosstab(
        df["hotel"],
        stay_categories,
        normalize="index"
    ).mul(100).round(2)
)

```

```python

cancellation_by_hotel = df.groupby("hotel").agg(
    total_bookings=("is_canceled", "size"),
    canceled_bookings=("is_canceled", "sum"),
    cancellation_rate=("is_canceled", "mean")
)

cancellation_by_hotel["cancellation_rate"] = (
    cancellation_by_hotel["cancellation_rate"] * 100
).round(2)

print(cancellation_by_hotel)
```

```python

status_by_hotel = pd.crosstab(
    df["hotel"],
    df["reservation_status"]
)

print("Reservation outcomes by hotel:")
print(status_by_hotel)

print("\nReservation outcome percentages by hotel:")
print(
    pd.crosstab(
        df["hotel"],
        df["reservation_status"],
        normalize="index"
    ).mul(100).round(2)
)
```

```python

channel_cancellations = df.groupby("distribution_channel").agg(
    total_bookings=("is_canceled", "size"),
    canceled_bookings=("is_canceled", "sum"),
    cancellation_rate=("is_canceled", "mean")
)

channel_cancellations["cancellation_rate"] = (
    channel_cancellations["cancellation_rate"] * 100
).round(2)

channel_cancellations = channel_cancellations.sort_values(
    "cancellation_rate",
    ascending=False
)

print(channel_cancellations)
```

```python

df["lead_time_category"] = pd.cut(
    df["lead_time"],
    bins=[-1, 7, 30, 90, 180, 365, float("inf")],
    labels=[
        "0–7 days",
        "8–30 days",
        "31–90 days",
        "91–180 days",
        "181–365 days",
        "More than 365 days"
    ]
)

lead_time_analysis = df.groupby(
    "lead_time_category",
    observed=False
).agg(
    total_bookings=("is_canceled", "size"),
    canceled_bookings=("is_canceled", "sum"),
    cancellation_rate=("is_canceled", "mean")
)

lead_time_analysis["cancellation_rate"] = (
    lead_time_analysis["cancellation_rate"] * 100
).round(2)

print(lead_time_analysis)
```

```python

deposit_analysis = df.groupby("deposit_type").agg(
    total_bookings=("is_canceled", "size"),
    canceled_bookings=("is_canceled", "sum"),
    cancellation_rate=("is_canceled", "mean")
)

deposit_analysis["cancellation_rate"] = (
    deposit_analysis["cancellation_rate"] * 100
).round(2)

print("Bookings and cancellation rates by deposit type:")
print(deposit_analysis.sort_values("cancellation_rate", ascending=False))

print("\nActual reservation outcomes by deposit type (%):")
print(
    pd.crosstab(
        df["deposit_type"],
        df["reservation_status"],
        normalize="index"
    ).mul(100).round(2)
)
```

```python
# Monthly bookings and actual cancellation rate

import pandas as pd

# Create arrival_date from the original date columns
df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str)
    + "-"
    + df["arrival_date_month"].astype(str)
    + "-"
    + df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d",
    errors="coerce"
)

# Summarize bookings and actual cancellations by month
monthly_bookings = (
    df.dropna(subset=["arrival_date"])
    .groupby(df["arrival_date"].dt.to_period("M"))
    .agg(
        total_bookings=("is_canceled", "size"),
        actual_cancellations=(
            "reservation_status",
            lambda s: (s == "Canceled").sum()
        )
    )
)

# Calculate actual cancellation rate
monthly_bookings["actual_cancellation_rate_pct"] = (
    monthly_bookings["actual_cancellations"]
    / monthly_bookings["total_bookings"]
    * 100
).round(2)

# Display results
print(monthly_bookings.to_string())

```

```python

# Phase 4 - Step 3: Monthly revenue trends

import pandas as pd
import matplotlib.pyplot as plt

# Keep completed stays only
completed = df[df["completed_stay_flag"]].copy()

# Construct arrival date from the separate date columns
completed["arrival_date"] = pd.to_datetime(
    completed["arrival_date_year"].astype(str)
    + "-"
    + completed["arrival_date_month"].astype(str)
    + "-"
    + completed["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d",
    errors="coerce"
)

# Remove rows where the date could not be constructed
completed = completed.dropna(subset=["arrival_date"])

# Group estimated room revenue by month and hotel
monthly_revenue = (
    completed
    .groupby([
        pd.Grouper(key="arrival_date", freq="MS"),
        "hotel"
    ])["estimated_room_revenue"]
    .sum()
    .reset_index()
    .sort_values(["arrival_date", "hotel"])
)

# Display the monthly revenue summary
display(monthly_revenue)

# Plot monthly revenue trends for each hotel
monthly_revenue_pivot = monthly_revenue.pivot(
    index="arrival_date",
    columns="hotel",
    values="estimated_room_revenue"
)

monthly_revenue_pivot.plot(
    figsize=(12, 6),
    marker="o"
)

plt.title("Monthly Estimated Room Revenue by Hotel")
plt.xlabel("Arrival Month")
plt.ylabel("Estimated Room Revenue")
plt.xticks(rotation=45)
plt.legend(title="Hotel Type")
plt.tight_layout()
plt.show()
```
